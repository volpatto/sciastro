## Método

Com $h=(b-a)/n$, usamos

$$T_n=h\left[\frac{f(a)+f(b)}{2}+\sum_{i=1}^{n-1}f(a+ih)\right].$$

O parâmetro $n$ é o número de subintervalos, não o número de pontos.


```python
def trapezoid(f, a, b, n):
    if not isinstance(n, int) or isinstance(n, bool) or n < 1:
        raise ValueError("n must be a positive integer")
    h = (b - a) / n
    interior = sum(f(a + i * h) for i in range(1, n))
    return h * (0.5 * f(a) + interior + 0.5 * f(b))
```

## Comparação com uma solução exata

Para $f(x)=x^2$ em $[0,1]$, a integral é $1/3$ e o erro em aritmética exata é

$$T_n-\frac{1}{3}=\frac{1}{6n^2}.$$

Por exemplo, com quatro subintervalos esperamos $T_4=0.34375$. A tolerância abaixo permite diferenças de arredondamento em ponto flutuante.


```python
from math import isclose

for n in [1, 2, 4, 8, 16]:
    approximation = trapezoid(lambda x: x * x, 0.0, 1.0, n)
    error = approximation - 1.0 / 3.0
    expected_error = 1.0 / (6 * n * n)
    assert isclose(error, expected_error, rel_tol=1e-10, abs_tol=1e-15)
    print(f"n={n:2d}  T_n={approximation:.10f}  error={error:.10f}")
```
